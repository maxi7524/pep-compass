```python
from pathlib import Path

project_root = Path("/home/prz/bioml/pep-compass")
```

```python
from Bio import SeqIO
import pandas as pd


def fasta_to_csv(directory_path, output_directory=None):
    """
    Convert all FASTA files in a directory to CSV format.

    Args:
        directory_path: Path to directory containing FASTA files
        output_directory: Path to directory where CSV files will be saved.
                         If None, saves to the same directory as the FASTA files.
    """
    directory = Path(directory_path)

    # Set output directory
    if output_directory is None:
        output_dir = directory
    else:
        output_dir = Path(output_directory)
        output_dir.mkdir(parents=True, exist_ok=True)

    # Find all .fasta files in the directory
    fasta_files = list(directory.glob("*.fasta"))

    print(f"Found {len(fasta_files)} FASTA files")

    for fasta_file in fasta_files:
        print(f"Processing {fasta_file.name}...")

        # Read sequences from FASTA file
        sequences = []
        for record in SeqIO.parse(fasta_file, "fasta"):
            sequences.append(
                {
                    "id": record.id,
                    "sequence": str(record.seq),
                    "description": record.description,
                }
            )

        # Convert to DataFrame
        df = pd.DataFrame(sequences)

        # Create output CSV filename (same name as FASTA but with .csv extension)
        csv_file = output_dir / fasta_file.with_suffix(".csv").name

        # Save to CSV
        df.to_csv(csv_file, index=False)

        print(f"  Saved {len(sequences)} sequences to {csv_file.name}")


# Example usage
dbaasp_dir = project_root / "data/hydramp/dbaasp"
fasta_to_csv(dbaasp_dir, dbaasp_dir / "csv")
```

```python
dbaasp_dir = project_root / "data/ampsphere"
fasta_to_csv(dbaasp_dir, dbaasp_dir / "csv")
```
