```python
# -*- coding: utf-8 -*-
# Amino acid classifications (1-letter codes, standard 20 proteinogenic amino acids)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

AMINO_ACID_CLASSES = {
    "polarity": {
        "hydrophobic": ["A", "V", "L", "I", "M", "F", "W", "P", "G"],
        "polar": ["S", "T", "Y", "C", "N", "Q"],
        "charged": ["K", "R", "H", "D", "E"],
    },

    "charge": {
        "positive": ["K", "R", "H"],
        "negative": ["D", "E"],
        "neutral": [
            "A", "V", "L", "I", "M", "F", "W", "P", "G",
            "S", "T", "Y", "C", "N", "Q"
        ],
    },

    "chemical_type": {
        "aliphatic": ["G", "A", "V", "L", "I"],
        "aromatic": ["F", "Y", "W"],
        "hydroxyl": ["S", "T", "Y"],
        "acidic": ["D", "E"],
        "amide": ["N", "Q"],
        "basic": ["K", "R", "H"],
        "sulfur": ["C", "M"],
        "imino": ["P"],
        # "other": [],  # placeholder (no gaps)
    },

    "essentiality": {
        "essential": ["F", "V", "T", "W", "M", "L", "I", "K", "H"],
        "non_essential": ["A", "N", "D", "E", "Q", "G", "P", "S", "C", "Y"],
    },

    "aromaticity": {
        "aromatic": ["F", "Y", "W"],
        "non_aromatic": [
            "A", "R", "N", "D", "C", "E", "Q", "G", "H", "I",
            "L", "K", "M", "P", "S", "T", "V"
        ],
    },

    "side_chain_size": {
        "small": ["G", "A", "S", "T", "P"],
        "medium": ["C", "N", "D", "Q", "E"],
        "large": ["V", "L", "I", "M", "F", "Y", "W", "K", "R", "H"],
    },
}
classes = ["polarity", "charge", "chemical_type", "essentiality", "aromaticity", "side_chain_size"]
# Just to be sure we always cover all 20 standard amino acids:
ALL_AA = set("ACDEFGHIKLMNPQRSTVWY")
for category, groups in AMINO_ACID_CLASSES.items():
    covered = set().union(*groups.values())
    missing = ALL_AA - covered
    if missing:
        groups["other_or_unclassified"] = sorted(list(missing))

ALL_AA = sorted("ACDEFGHIKLMNPQRSTVWY")

```

```python
def create_square_plot(
    data: np.ndarray,
    classification: str | None = None,
    cmap: str = "coolwarm",
    title: str | None = None,
    show_labels: bool = True,
    figsize=(10, 8),
    return_fig: bool | None = None,
    ax: plt.Axes | None = None,   
):
    """
    Create a sorted amino acid × amino acid heatmap based on a chosen classification.

    If `ax` is provided, the plot will be drawn on that axis instead of creating a new figure.
    """

    data = np.array(data)
    n = data.shape[0]
    if data.shape[0] != data.shape[1]:
        raise ValueError("Input data must be a square matrix!")

    aas = ALL_AA if n == 20 else [f"AA{i+1}" for i in range(n)]
    df = pd.DataFrame(data, index=aas, columns=aas)

    # Create figure/axes if not provided
    if ax is None:
        fig, ax = plt.subplots(figsize=figsize)
    else:
        fig = plt.gcf()

    # === Case 1: no classification ===
    if classification is None:
        sns.heatmap(df, cmap=cmap, square=True, cbar=True, ax=ax)
        if not title:
            title = "Amino acid × amino acid (unsorted)"
        ax.set_title(title, fontsize=14)
        ax.set_xlabel("Amino acid")
        ax.set_ylabel("Amino acid")
        plt.tight_layout()
        if return_fig:
            return fig, ax
        else:
            return None

    # === Case 2: classification selected ===
    if classification not in AMINO_ACID_CLASSES:
        raise ValueError(f"Unknown classification: {classification}")

    groups = AMINO_ACID_CLASSES[classification]
    sorted_groups = list(groups.keys())

    sorted_aas = [aa for group in sorted_groups for aa in aas if aa in groups[group]]
    missing = [aa for aa in aas if aa not in sorted_aas]
    sorted_aas.extend(missing)

    df_sorted = df.loc[sorted_aas, sorted_aas]

    sns.heatmap(df_sorted, cmap=cmap, square=True, cbar=True, ax=ax)

    if not title:
        title = f"Amino acid × amino acid ({classification} classification)"
    ax.set_title(title, fontsize=14)
    ax.set_xlabel("Amino acid")
    ax.set_ylabel("Amino acid")

    # === Group boundaries and labels ===
    group_sizes = [len(groups[g]) for g in sorted_groups if any(aa in aas for aa in groups[g])]
    boundaries = np.cumsum(group_sizes)

    for b in boundaries[:-1]:
        ax.axhline(b, color="black", lw=1)
        ax.axvline(b, color="black", lw=1)

    if show_labels:
        midpoints = np.array(boundaries) - np.array(group_sizes) / 2
        for i, g in enumerate(sorted_groups):
            mid = midpoints[i]
            ax.text(-1.5, mid, g, va="center", ha="right", fontsize=10)
        ylim = ax.get_ylim()
        xlim = ax.get_xlim()
        offset = len(aas) * 0.05
        for i, g in enumerate(sorted_groups):
            mid = midpoints[i]
            ax.text(
                mid,
                len(aas) + offset,
                g,
                va="top",
                ha="center",
                fontsize=10,
            )
        ax.set_ylim(ylim)
        ax.set_xlim(xlim)

    plt.tight_layout()
    if return_fig:
        return fig, ax
    else:
        return None
```

```python
import numpy as np
import pandas as pd

matrix = pd.DataFrame(
    np.nan,  # możesz też dać np.zeros(...)
    index=ALL_AA,
    columns=ALL_AA
)
```

```python
for (aa1, aa2), value in bootstrap_results.items():
    if aa1 in ALL_AA and aa2 in ALL_AA:
        matrix.loc[aa1, aa2] = (value['E. coli ATCC 11775_diff_rank'] + value['E. coli AIG221_diff_rank'] + value['E. coli AIG222_diff_rank']) / 3.0  # average over three strains

matrix = matrix.fillna(0)
create_square_plot(
    matrix,
    classification="charge",
    cmap="coolwarm",
    # vmin=-0.1,
    # vmax=0.1,
    show_labels=True
)
```

```python
metric_names = list(next(iter(bootstrap_results.values())).keys())

# Wybierz konkretne metryki po indeksach
selected_indices = [3, 10, 15, 20]
selected_metrics = [metric_names[i] for i in selected_indices if i < len(metric_names)]

# Utwórz figure 2x2
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
axes = axes.flatten()

for ax, metric in zip(axes, selected_metrics):
    
    matrix = pd.DataFrame(np.nan, index=ALL_AA, columns=ALL_AA)

    
    for (aa1, aa2), value in bootstrap_results.items():
        if aa1 in ALL_AA and aa2 in ALL_AA:
            matrix.loc[aa1, aa2] = value[metric]

    matrix = matrix.fillna(0)

    
    create_square_plot(
        matrix,
        classification="charge",   # możesz zmienić np. na "polarity"
        cmap="coolwarm",
        title=metric,              # nazwa metryki jako tytuł
        show_labels=True,
        ax=ax,
        return_fig=False
    )

# 4️⃣ Dopasuj layout i pokaż
plt.tight_layout()
plt.show()
```

```python
metric_names = list(next(iter(bootstrap_results.values())).keys())

# Podział na grupy
gram_neg = [0,1,2,3,4,5,6,11,12,13,16,17,18,19,20,22,23,24,29,30,31,32]
gram_pos = [7,8,9,10,14,15,21,25,26,27,28,33]

gram_neg_metrics = [metric_names[i] for i in gram_neg if i < len(metric_names)]
gram_pos_metrics = [metric_names[i] for i in gram_pos if i < len(metric_names)]

# Funkcja pomocnicza: oblicz średnią macierz dla danej grupy metryk
def compute_mean_matrix(metrics):
    matrix = pd.DataFrame(np.nan, index=ALL_AA, columns=ALL_AA)
    for (aa1, aa2), values in bootstrap_results.items():
        if aa1 in ALL_AA and aa2 in ALL_AA:
            vals = [values[m] for m in metrics if m in values]
            if len(vals) > 0:
                matrix.loc[aa1, aa2] = np.mean(vals)
    return matrix.fillna(0)

# Oblicz średnie macierze
mean_gram_neg_matrix = compute_mean_matrix(gram_neg_metrics)
mean_gram_pos_matrix = compute_mean_matrix(gram_pos_metrics)

# === Wizualizacja obok siebie ===
fig, axes = plt.subplots(1, 2, figsize=(18, 8))

create_square_plot(
    mean_gram_neg_matrix,
    classification="charge",
    cmap="coolwarm",
    title="Gram− mean rank",
    show_labels=True,
    ax=axes[0],
    return_fig=False,
)

create_square_plot(
    mean_gram_pos_matrix,
    classification="charge",
    cmap="coolwarm",
    title="Gram+ mean rank",
    show_labels=True,
    ax=axes[1],
    return_fig=False,
)

fig.suptitle("Average rank for Gram− vs Gram+ metrics", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

```python
diff_matrix = mean_gram_neg_matrix - mean_gram_pos_matrix

create_square_plot(
    diff_matrix,
    classification="polarity",
    cmap="coolwarm",
    title="Difference in mean rank (Gram− minus Gram+)",
    show_labels=True,
    figsize=(10, 8),
)
```
