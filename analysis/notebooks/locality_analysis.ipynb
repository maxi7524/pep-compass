# Locality experiment analysis
This notebook indexes manifests first and reads large tracking CSVs only for the selected analysis.

```python
from pathlib import Path
from pep_compass.experiments.analysis import LocalityExperiment
from pep_compass.experiments.analysis.methods import evaluation_summary, filter_score_summary, reconstruct_trajectory, retention_summary, stratified_candidate_sample

RESULTS = Path('../../results/locality').resolve()
experiment = LocalityExperiment.open(RESULTS)
experiment.catalog[['experiment', 'method', 'name', 'repetition', 'tracking_level']].head()
```

```python
selection = experiment.select(iteration_min=0)
evaluation_summary(selection, ['experiment', 'method'])
```

```python
retention_summary(selection, ['experiment', 'method'])
```

```python
filter_score_summary(selection, ['experiment', 'method'])
```

```python
# Inspect one trajectory without loading candidates from other runs.
row = selection.runs.iloc[0]
reconstruct_trajectory(selection, row.task_id, iteration_id=1, trajectory_id=0)
```

```python
# Bounded PoGS input: equal reservoir samples per method/parent/grid/status/Hamming stratum.
pogs_selection = experiment.select(methods=['lams', 'random_walker'])
pogs_sample = stratified_candidate_sample(pogs_selection, candidates_per_stratum=25)
pogs_sample.head()
```
