```python
import os
import pandas as pd

def cmaes_get_proper_datasets_max(peptide_name, folder, revert_sign=False):
    """
    Dla podanego peptydu znajduje wszystkie pliki CSV zaczynające się od peptide_name
    i zwraca listę maksymalnych wartości z kolumny 'score' na podstawie wierszy 1 i 2.
    """

    best_scores = []

    for filename in os.listdir(folder):
        if filename.startswith(peptide_name) and filename.endswith(".csv"):
            filepath = os.path.join(folder, filename)

            try:
                # wczytanie tylko wierszy 1 i 2
                df = pd.read_csv(
                    filepath,
                    sep=",",
                    header=None,
                    skiprows=lambda i: i ==0
                )

                # wybieramy kolumny 1 i 2
                df = df.iloc[:, [1, 2]]


                df.columns = ['sequence','score']

                # konwertujemy score na float (na wypadek)
                df["score"] = pd.to_numeric(df["score"], errors="coerce")

                # bierzemy max
                max_val = df["score"].max()

                if revert_sign:
                    max_val = -max_val

                best_scores.append(max_val)

            except Exception as e:
                print(f"Nie mogę wczytać pliku {filename}: {e}")
                continue

    return best_scores
```

```python
import os
import pandas as pd

def cmaes_get_proper_datasets_max_2(peptide_name, folder, revert_sign=False):
    """
    For a given peptide_name finds CSV files starting with peptide_name.
    For each file reads rows 1 and 2, takes columns 'sequence' and 'score',
    returns:
        - best_scores: list of max score values
        - best_sequences: list of sequences that achieved the max score
    """

    best_scores = []
    best_sequences = []

    for filename in os.listdir(folder):
        if filename.startswith(peptide_name) and filename.endswith(".csv"):
            filepath = os.path.join(folder, filename)

            try:
                # load rows except the header, but keep row indices matched
                df = pd.read_csv(
                    filepath,
                    sep=",",
                    header=None,
                    skiprows=lambda i: i == 0
                )

                # select columns 1 (sequence) and 2 (score)
                df = df.iloc[:, [1, 2]]
                df.columns = ['sequence', 'score']

                # numeric score conversion
                df["score"] = pd.to_numeric(df["score"], errors="coerce")

                # compute max
                max_val = df["score"].max()

                # optionally revert sign
                if revert_sign:
                    true_max = -max_val
                else:
                    true_max = max_val

                best_scores.append(true_max)

                # extract the sequence corresponding to original max_val
                best_seq = df.loc[df["score"] == max_val, "sequence"].iloc[0]
                best_sequences.append(best_seq)

            except Exception as e:
                print(f"Could not load {filename}: {e}")
                continue

    return best_scores, best_sequences
```

```python
def create_table_row_2(peptides, names, base_folder, revert_sign=False):
    """
    Create a table with n_runs rows.
    Each column is a peptide (name), and each cell contains:
        sequence (score)
    for a particular run.
    """

    # First column processed determines n_runs
    all_columns = {}

    max_rows = 0

    for peptide, name in zip(peptides, names):
        folder = os.path.join(base_folder)

        scores, sequences = cmaes_get_proper_datasets_max_2(peptide, folder, revert_sign=revert_sign)

        if scores and sequences:
            # Convert to "sequence (score)" strings
            column_cells = [f"{seq} ({score:.3f})" for seq, score in zip(sequences, scores)]
        else:
            column_cells = ["No data"]

        all_columns[name] = column_cells
        max_rows = max(max_rows, len(column_cells))

    # Normalize column lengths (pad with empty strings)
    for name in all_columns:
        col = all_columns[name]
        if len(col) < max_rows:
            col += [""] * (max_rows - len(col))
        all_columns[name] = col

    # Create DataFrame with n_runs rows
    df = pd.DataFrame(all_columns)
    return df
```

```python
def create_comprehensive_table_with_sequences():
    """
    Create a comprehensive table showing scores and sequences for all methods
    """
    
    methods_config = {
        'CMA-ES (Hydrophobicity)': {
            'folder': '/home/kjurasz/pep-compass/results/cma_es_hydrophobicity/HydrAMPWrappedHydrophobicity',
            'revert_sign': False
        },
        'CMA-ES (BattleAMP)': {
            'folder': '/home/kjurasz/pep-compass/results/cma_es/HydrAMPWrappedBattleAMP', 
            'revert_sign': True
        },
        'CMA-ES (ToxiPep)': {
            'folder': '/home/kjurasz/pep-compass/results/cma_es/HydrAMPWrappedToxiPep',
            'revert_sign': True
        },
        'Random Mutation (Hydrophobicity)': {
            'folder': '/home/kjurasz/pep-compass/results/random_mutation/Hydrophobicity',
            'revert_sign': False
        },
        'Random Mutation (BattleAMP)': {
            'folder': '/home/kjurasz/pep-compass/results/random_mutation/BattleAMP',
            'revert_sign': True
        },
        'Random Mutation (ToxiPep)': {
            'folder': '/home/kjurasz/pep-compass/results/random_mutation/ToxiPep',
            'revert_sign': True
        }
    }
    
    all_rows = []
    
    for method_name, config in methods_config.items():
        print(f"\nProcessing {method_name}...")
        
        try:
            row = create_table_row_2(
                peptides, 
                names, 
                config['folder'], 
                revert_sign=config['revert_sign']
            )
            row.index = [method_name]
            all_rows.append(row)
            print(f"✅ Successfully processed {method_name}")
            
        except Exception as e:
            print(f"❌ Error processing {method_name}: {e}")
            # Create empty row
            empty_row = {name: "Error" for name in names}
            row = pd.DataFrame([empty_row], index=[method_name])
            all_rows.append(row)
    
    # Combine all rows
    if all_rows:
        combined_table = pd.concat(all_rows)
        
        print(f"\n{'='*80}")
        print("COMPREHENSIVE RESULTS TABLE WITH SEQUENCES")
        print(f"{'='*80}")
        
        # Save to CSV
        output_file = "comprehensive_results_with_sequences.csv"
        combined_table.to_csv(output_file)
        print(f"💾 Saved to: {output_file}")
        
        return combined_table
    else:
        return None
```

```python
def get_mean_sd(scores):
    import numpy as np
    mean_score = np.mean(scores)
    sd_score = np.std(scores)
    return mean_score, sd_score


def create_table_row(peptides, names, base_folder, revert_sign=False):
    row_dict = {}

    for peptide, name in zip(peptides, names):
        folder = os.path.join(base_folder)

        scores = cmaes_get_proper_datasets_max(peptide, folder, revert_sign=revert_sign)
        mean_score, sd_score = get_mean_sd(scores)

        # każda kolumna jest nazwana nazwą peptydu
        row_dict[name] = f"{mean_score:.2f} ± {sd_score:.3f}"

    # zwracamy 1-wierszowy DataFrame
    df = pd.DataFrame([row_dict])
    return df
```
