```python
import pandas as pd
```

```python
from pathlib import Path

dir_path = Path('/home/adambiel/pep-compass/results/lebo/Hydrophobicity')
csv_files = sorted(dir_path.glob('*.csv'))

data = []

# time,sequence,score,latent_point

for csv_file in csv_files:
    df = pd.read_csv(csv_file)
    
    if len(df) < 1400:
        continue
    
    starting_peptide = str(csv_file).split('/')[-1].split('_')[0]
    
    data.append({
        "starting_peptide": starting_peptide,
        "the_best_peptide": df.loc[df['score'].idxmax()]['sequence'],
        "the_best_score": df['score'].max(),
        "scores": df['score'].tolist(),
        "peptides": df['sequence'].tolist()
    })
    
summary_df = pd.DataFrame(data)
summary_df[["starting_peptide", "the_best_peptide", "the_best_score"]]
```
